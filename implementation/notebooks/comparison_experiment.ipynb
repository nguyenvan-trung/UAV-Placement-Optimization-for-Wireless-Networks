## Chạy thí nghiệm
Thay phần dữ liệu minh họa bằng lời gọi optimizer sau khi hoàn thiện thuật toán.

```python
# TODO: kết quả thật sẽ được tạo bởi GAOptimizer, PSOOptimizer và WOAOptimizer.
results = pd.DataFrame(columns=[
    'algorithm', 'seed', 'coverage', 'best_objective',
    'iterations_to_converge', 'runtime_seconds'
])
results
```

## Bảng tổng hợp

```python
if results.empty:
    display(pd.DataFrame({'Thông báo': ['Chưa có kết quả; cần hoàn thiện các optimizer.']}))
else:
    summary = results.groupby('algorithm').agg(
        coverage_mean=('coverage', 'mean'),
        objective_mean=('best_objective', 'mean'),
        convergence_mean=('iterations_to_converge', 'mean'),
        runtime_mean=('runtime_seconds', 'mean'),
    )
    display(summary)
```

## Biểu đồ so sánh

```python
if not results.empty:
    results.groupby('algorithm')['best_objective'].mean().plot(kind='bar')
    plt.title('So sánh giá trị hàm mục tiêu')
    plt.ylabel('Objective value')
    plt.show()
```
